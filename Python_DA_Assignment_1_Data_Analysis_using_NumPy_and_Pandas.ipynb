```python
# Part 1: NumPy Array Operations

# Step 1: Create a 1D NumPy Array
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:", temperatures_w1)
```

np.array() creates a NumPy array containing the temperatures recorded during Week 1.

```python
# Step 2: Inspect Array Properties
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```



*   shape returns the dimensions of the array
*   dtype returns the data type of its elements
*   size returns the total number of elements






```python
# Step 3: Perform Array Operations
# Convert Celsius to Fahrenheit
fahrenheit = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", np.round(fahrenheit, 2))

# Calculate maximum, minimum, and mean
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", round(np.mean(temperatures_w1), 2))
```



*   The formula converts each Celsius value into Fahrenheit
*   np.max() finds the highest temperature
*   np.min() finds the lowest temperature
*   np.mean() calculates the average temperature



```python
# Step 4: Array Slicing and Indexing
# First three days
print("First Three Days:", temperatures_w1[:3])

# Weekend (last two days)
print("Weekend Temperatures:", temperatures_w1[-2:])

# Middle three days
print("Middle Three Days:", temperatures_w1[2:5])
```



*   [:3] extracts the first three elements
*   [-2:] extracts the last two elements
*   [2:5] extracts elements at index positions 2, 3, and 4




```python
# Step 5: Create a 2D NumPy Array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperatures for Two Weeks:")
print(temperatures)
```

A 2D array contains rows and columns. Here, each row represents one week, and each column represents a day.

```python
# Step 6: Inspect and Slice the 2D Array
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)

# Extract each week
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])

# Extract weekends for both weeks
print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```



*   shape returns 2 rows and 7 columns
*   size returns 14 elements
*   temperatures[0] selects Week 1
*   temperatures[1] selects Week 2
*   temperatures[0, -2:] selects the last two temperatures of Week 1




```python
# Part 2: Pandas Series

# Step 1: Create a Pandas Series
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

pd.Series() creates a one-dimensional labeled data structure. The custom indices identify each student's rank.

```python
# Step 2: Indexing and Slicing
# Access the first rank using integer position
print("1st Rank Mark:", marks.iloc[0])

# Retrieve the top three ranks using loc
print("Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# Retrieve the third rank using iloc
print("3rd Rank Mark:", marks.iloc[2])

# Filter marks greater than 90
print("Marks Greater Than 90:")
print(marks[marks > 90])
```



*   iloc[0] accesses the first element by position
*   loc[] retrieves elements using index labels
*   iloc[2] accesses the third element
*   marks > 90 creates a Boolean condition to filter marks greater than 90



```python
# Step 3: Manipulate the Series
# Modify the first rank mark
marks.loc['Rank1'] = 100
print("Updated Marks:")
print(marks)

# Remove the last rank
marks = marks.drop('Rank5')
print("After Removing Last Rank:")
print(marks)

# Calculate CGPA
cgpa = marks / 10
print("CGPA:")
print(cgpa)
```



*   marks.loc['Rank1'] = 100 updates the first rank's mark
*   drop('Rank5') removes the last rank from the Series
*   marks / 10 divides each remaining mark by 10 to calculate CGPA




```python
# Part 3: Pandas DataFrame

# Step 1: Create a Pandas DataFrame
import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print(transactions)
```

pd.DataFrame() creates a two-dimensional table with rows and columns using the supplied dictionary.

```python
# Step 2: Explore the DataFrame
# A.Display the DataFrame and basic information
print("Complete DataFrame:")
print(transactions)

print("\nFirst 5 Rows:")
print(transactions.head())

print("\nLast 5 Rows:")
print(transactions.tail())

print("\nShape:")
print(transactions.shape)

print("\nColumn Names:")
print(transactions.columns)

print("\nData Types:")
print(transactions.dtypes)

print("\nBasic Information:")
transactions.info()
```



*   head() displays the first 5 rows
*   tail() displays the last 5 rows
*   shape returns (10, 4)
*   columns displays the four column names
*   dtypes displays the data type of each column
*   info() displays column details, non-null counts, and memory usage



```python
# B.Display only ProductCategory and Amount
print(transactions[['ProductCategory', 'Amount']])
```

Double square brackets select multiple columns from a DataFrame.

```python
# C.Retrieve the last 3 columns using iloc
print(transactions.iloc[:, -3:])
```



*   Displays the ProductCategory, Region, and Amount columns for all 10 transaction
*   selects all rows, while -3: selects the last three columns



```python
# D.Filter North region transactions where Amount is greater than 200
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(filtered_data)
```

The & operator combines both conditions. Only rows matching both conditions are returned.

```python
# E.Find value counts for ProductCategory
print(transactions['ProductCategory'].value_counts())
```

value_counts() counts how many times each product category appears.

```python
# F.Find unique values in Region
print(transactions['Region'].unique())
```

unique() returns the distinct region values without repeating them.

```python
# G.Group by Region and calculate the mean Amount
print(transactions.groupby('Region')['Amount'].mean())
```

groupby() groups transactions by region, and mean() calculates the average transaction amount for each region.

```python
# Step 3: Manipulate the DataFrame

# A.Modify Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print(transactions)
```

loc[] identifies the row where TransactionID is 102 and updates its Amount to 165.

```python
# B.Add a Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

The new column stores 10% of each transaction amount. For example, 10% of 200 is 20.

```python
# C.Remove the row with TransactionID 109
transactions = transactions[
    transactions['TransactionID'] != 109
]

print(transactions)
```

The condition TransactionID != 109 keeps every row except the row with TransactionID 109.

```python
# D.Delete the Discount column
transactions = transactions.drop(columns=['Discount'])

print(transactions)
```

drop(columns=['Discount']) removes the Discount column. The final DataFrame contains 9 rows and 4 columns.
